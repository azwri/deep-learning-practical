```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./data/used_cars.csv")
```

```python
price = df['price']
price = price.str.replace('$', '')
price = price.str.replace(',', '')
price = price.astype(int)

print(price)
```

```python
price[price < 300000].hist(bins=30)
```

```python
df['price'] = price
df.head()
```

```python
df['price'].mean()
```

```python
df_hyundai = df[df['brand'] == 'Hyundai']
df_hyundai['price'].mean()
```

```python
df_ford = df[df['brand'] == 'Ford']
df_ford['price'].mean()
```

```python
df[['model_year', 'price']].corr()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['model_year'], df['price'])
```

```python
df['age'] = df['model_year'].max() - df['model_year']
df.head()
```

```python
milage = df['milage']
milage = milage.str.replace(',', '')
milage = milage.str.replace('mi.', '')
milage = milage.astype(int)
print(milage)
```

```python
df['milage'] = milage
df.head()
```

```python
df.to_csv("./data/used_cars_clean.csv", index=False)
```
