```python
import matplotlib.pyplot as plt
import torch
from torch import nn
```

```python
nn.functional.sigmoid(input=torch.tensor([-100, -0.1,  1, -1, 2, 3]))
```

```python
X = torch.arange(-5, 5, 0.1)
y = nn.functional.sigmoid(X)
print(X, y, sep='\n===================\n')
```

```python
plt.plot(X, y)
```
