```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
# test
cv = CountVectorizer(max_features=6)
vectorizer = TfidfVectorizer(max_features=6)
documents = [
    'Hello, today is amazing. Hello, hello',
    'Hello, today is perfect',
]
out = cv.fit_transform(raw_documents=documents)
out_ = vectorizer.fit_transform(raw_documents=documents)
print(out.todense())
print(cv.get_feature_names_out())
print("="*30)
print(out_.todense())
print(vectorizer.get_feature_names_out())
print("="*30)
print(cv.inverse_transform(np.array(out.todense()[0])))
print(np.array(out.todense()[0]))
print("="*30)
```

```python
df = pd.read_csv('./data/SMSSpamCollection', sep='\t', names=['type', 'message'])
df['spam'] = df['type'] == 'spam'
df = df.drop(columns=['type'])
print(df['spam'].value_counts())
print("="*30)
print(df.info())
```

```python
cv = CountVectorizer(max_features=1000)
messages = cv.fit_transform(df['message'])
# print(messages)
print(messages[0])
print(messages.todense()[0])
```
